```python
from pathlib import Path

# 1. Define your folder path and the name of your final merged file
folder_path = Path("C:/Users/ezra/OneDrive - Delft University of Technology/School/Universiteit/BEP/Raw Data/Fragmented 07-10")
output_file = Path("merged_logs.csv")

# 2. Find all log files (assuming they end in .txt based on your upload)
# Change "*.txt" to "*.csv" if your files use a different extension
file_list = list(folder_path.glob("*.txt"))

# 3. Open the main output file in write mode
with open(output_file, 'w', encoding='utf-8') as outfile:
    
    # 4. Loop through every file found in the folder
    for i, file_path in enumerate(file_list):
        with open(file_path, 'r', encoding='utf-8') as infile:
            
            # Read the very first line of the current file (the header)
            header = infile.readline()
            
            # If this is the very first file, write the header to the output file
            if i == 0:
                outfile.write(header)
                
            # Now loop through the remaining lines of data and write them
            for line in infile:
                outfile.write(line)
                
            # Add a safety newline character at the end of each file's data
            # This prevents the last line of File A from merging with the first line of File B
            if not line.endswith('\n'):
                outfile.write('\n')

print(f"Successfully merged {len(file_list)} files into {output_file.name}")
```
