```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def Molarsat(T): #saturation molarity at T[C] (Arden Buck for T>0)
    R = 8.314
    return 611.21/R/(T+273.15)*np.exp((18.678 - T/234.5 )*( T/(257.14 + T)))
    
def Absorption_coef(M_s,alpha_h2o,x,RH): #absorption coefficient in a natural exponential
    return M_s*alpha_h2o*x*RH

T = 35
M_s = Molarsat(T)
alpha_h2o = 2/(1e5/8.314/296)
x = 1
RH = 1
I_scaled = np.exp(-Absorption_coef(M_s,alpha_h2o,x,RH))
print(np.exp(-Absorption_coef(M_s,alpha_h2o,x,RH)))

Tmin = 0
Tmax = 70
T_arr = np.linspace(Tmin,Tmax,1000)
Msat_arr = Molarsat(T_arr)

plt.figure()
plt.plot(T_arr,Msat_arr,linewidth=2)
plt.xlabel(f'Air temperature [\N{DEGREE SIGN}C]')
plt.ylabel(r'Saturation molarity [mol/m$^3$]')
plt.grid()
plt.savefig('Saturation molarity', dpi = 300)
plt.ylim(0,np.max(Msat_arr))
plt.xlim(Tmin,Tmax)
plt.show()
```
